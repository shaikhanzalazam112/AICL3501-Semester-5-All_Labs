## Part 7 — Tensor Operations
Tensor operations mirror NumPy operations (element-wise add, matrix multiply, transpose, reduce sum, reduce mean).

```python
# Tensor operations mirror NumPy
t_a = tf.constant([[1., 2.], [3., 4.]])
t_b = tf.constant([[5., 6.], [7., 8.]])

print("Element-wise add:\n", t_a + t_b)
print("\nMatrix multiply:\n", tf.matmul(t_a, t_b))
print("\nTranspose:\n", tf.transpose(t_a))
print("\nReduce sum:", tf.reduce_sum(t_a))
print("Reduce mean along axis 0:", tf.reduce_mean(t_a, axis=0))
```

## Part 8 — Toy Regression Dataset
Generate a toy regression dataset: y = 3x + 2 + noise, and visualize it with Matplotlib.

```python
# Generate a toy regression dataset: y = 3x + 2 + noise
np.random.seed(0)
X = np.linspace(-5, 5, 200).reshape(-1, 1).astype("float32")
y_true = 3 * X + 2 + np.random.normal(0, 1, X.shape).astype("float32")

plt.figure(figsize=(6, 4))
plt.scatter(X, y_true, alpha=0.6, s=15)
plt.title("Toy dataset: y = 3x + 2 + noise")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

## Part 9 — Original Keras Model
Building neural networks with Keras. Using the Sequential API, layers are stacked in order, the model is compiled with an optimizer and a loss function, and then trained (`fit`) on data.

Build a simple Sequential model (one dense layer = linear regression).

```python
# Build a simple Sequential model (one dense layer = linear regression)
model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(1,))
])

model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.01),
              loss="mse")

model.summary()
```

## Part 10 — Original Training and Results
Train the model, plot the training loss over epochs, visualize the fitted line against the data, and print the learned weight and bias.

```python
# Train the model
history = model.fit(X, y_true, epochs=50, verbose=0)

plt.figure(figsize=(6, 4))
plt.plot(history.history["loss"])
plt.title("Training loss over epochs")
plt.xlabel("epoch")
plt.ylabel("MSE loss")
plt.show()
```

```python
# Visualize the fitted line against the data
y_pred = model.predict(X, verbose=0)

plt.figure(figsize=(6, 4))
plt.scatter(X, y_true, alpha=0.5, s=15, label="data")
plt.plot(X, y_pred, color="red", linewidth=2, label="model fit")
plt.title("Fitted linear model")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
learned_weight, learned_bias = model.layers[0].get_weights()
print(f"Learned: y = {learned_weight[0][0]:.2f}x + {learned_bias[0]:.2f}")
print("True relationship: y = 3x + 2")
```

## Part 11 — Modified Keras Model
### Modified Model — Optimizer and Loss Function

This section creates a **separate** model using the same dataset and architecture as the original, but with a different optimizer and loss function:
- Optimizer: **Adam** (instead of SGD)
- Learning rate: **0.01** (unchanged)
- Loss: **MAE** (instead of MSE)
- Epochs: **50** (unchanged)

The original model and its results above are **not modified**.

```python
# (a) Modified model creation
modified_model = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(1,))
])

modified_model.summary()
```

```python
# (b) Modified model compilation
modified_model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
                        loss="mae")
```

## Part 12 — Modified Training and Results

```python
# (c) Modified model training
modified_history = modified_model.fit(X, y_true, epochs=50, verbose=0)
```

```python
# (d) Modified training-loss graph
plt.figure(figsize=(6, 4))
plt.plot(modified_history.history["loss"], color="green")
plt.title("Modified model: Training loss over epochs (Adam + MAE)")
plt.xlabel("epoch")
plt.ylabel("MAE loss")
plt.show()
```

```python
# (e) Modified fitted-model graph
y_pred_modified = modified_model.predict(X, verbose=0)

plt.figure(figsize=(6, 4))
plt.scatter(X, y_true, alpha=0.5, s=15, label="data")
plt.plot(X, y_pred_modified, color="orange", linewidth=2, label="modified model fit")
plt.title("Modified Fitted Linear Model (Adam + MAE)")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.show()
```

```python
# (f) Learned weight/bias output
learned_weight_mod, learned_bias_mod = modified_model.layers[0].get_weights()
print(f"Modified model learned: y = {learned_weight_mod[0][0]:.2f}x + {learned_bias_mod[0]:.2f}")
print("True relationship: y = 3x + 2")
```

```python
print("=== Original Model (SGD + MSE) ===")
print(f"Final training loss: {history.history['loss'][-1]:.4f}")
print(f"Learned weight: {learned_weight[0][0]:.4f}, Learned bias: {learned_bias[0]:.4f}")

print("\n=== Modified Model (Adam + MAE) ===")
print(f"Final training loss: {modified_history.history['loss'][-1]:.4f}")
print(f"Learned weight: {learned_weight_mod[0][0]:.4f}, Learned bias: {learned_bias_mod[0]:.4f}")

print("\nTrue relationship: y = 3x + 2")
```

### Discussion

- **Training behavior:** The original SGD + MSE model shows a smooth, steadily decreasing loss curve typical of gradient descent on a convex mean-squared-error surface. The modified Adam + MAE model converges using Adam's adaptive per-parameter learning rates, and because MAE penalizes errors linearly (not quadratically), its loss curve looks different in shape and scale from the MSE curve, so the two loss values are not directly comparable on the same axis.
- **Final loss:** The two loss numbers are printed above; they are on different scales (MSE vs. MAE) so a smaller/larger number does not by itself mean "better" — what matters is how close the learned weight and bias are to the true relationship y = 3x + 2.
- **Learned weight/slope and bias/intercept:** Both models are compared above against the true slope (3) and intercept (2). Whichever model's learned weight and bias are closer to 3 and 2 respectively fit the underlying relationship more accurately.
- **Fitted line:** Both fitted-line plots (original and modified) show a straight line through the scattered data; visually comparing the two plots shows how closely each configuration recovers the true linear trend.
- **Overall result:** Both optimizer/loss combinations are able to learn a reasonable approximation of the linear relationship given 50 epochs on this simple one-feature dataset. Adam typically converges faster in early epochs due to adaptive learning rates, while SGD with a fixed learning rate converges more gradually. Since MAE is less sensitive to large deviations than MSE, the modified model's fit can be slightly more robust to any outlier noise in the toy dataset, but for this simple, mostly linear, low-noise dataset the two configurations are expected to perform **similarly** overall, with only minor differences in the exact learned weight and bias.
